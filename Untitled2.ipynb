```python
import numpy as np
import pandas as pd
import os

def generate_printing_data(num_samples=250, version=1, seed=42):
    np.random.seed(seed)

    # Feature 1: Material
    materials = np.random.choice(['PLA', 'ABS', 'PETG'], size=num_samples, p=[0.5, 0.3, 0.2])

    # Features depending on material
    nozzle_temp = []
    bed_temp = []
    print_speed = []

    for mat in materials:
        if mat == 'PLA':
            nozzle = np.random.normal(205, 5)
            # Version 2 has reduced bed temperatures
            bed = np.random.normal(55, 3) if version == 1 else np.random.normal(45, 3)
            # Version 2 has increased printing speeds
            speed = np.random.normal(50, 5) if version == 1 else np.random.normal(70, 5)
        elif mat == 'ABS':
            nozzle = np.random.normal(240, 5)
            bed = np.random.normal(100, 5) if version == 1 else np.random.normal(85, 5)
            speed = np.random.normal(40, 5) if version == 1 else np.random.normal(60, 5)
        else: # PETG
            nozzle = np.random.normal(230, 5)
            bed = np.random.normal(75, 4) if version == 1 else np.random.normal(65, 4)
            speed = np.random.normal(45, 5) if version == 1 else np.random.normal(65, 5)

        nozzle_temp.append(nozzle)
        bed_temp.append(bed)
        print_speed.append(speed)

    ambient_temp = np.random.normal(23, 2, size=num_samples)
    cooling_fan = np.random.choice([0, 20, 50, 100], size=num_samples, p=[0.2, 0.2, 0.3, 0.3])
    base_area = np.random.uniform(10, 250, size=num_samples) # mm^2
    layer_height = np.random.choice([0.12, 0.2, 0.28], size=num_samples, p=[0.3, 0.5, 0.2])

    df = pd.DataFrame({
        'material': materials,
        'nozzle_temperature': nozzle_temp,
        'bed_temperature': bed_temp,
        'printing_speed': print_speed,
        'ambient_temperature': ambient_temp,
        'cooling_fan_speed': cooling_fan,
        'component_base_area': base_area,
        'layer_height': layer_height
    })

    # Target rules (Probability of warping)
    # Warping happens if: Bed is too cold for ABS, print speed is too high, cooling is too high on ABS, base area is too large
    warping_prob = []
    for idx, row in df.iterrows():
        p = 0.05
        if row['material'] == 'ABS':
            if row['bed_temperature'] < 90: p += 0.4
            if row['cooling_fan_speed'] > 20: p += 0.3
        if row['material'] == 'PLA' and row['bed_temperature'] < 50: p += 0.2
        if row['printing_speed'] > 60: p += 0.25
        if row['component_base_area'] > 180: p += 0.15

        p = min(max(p, 0.0), 0.95)
        warping_prob.append(p)

    df['warping_defect'] = [1 if np.random.rand() < p_val else 0 for p_val in warping_prob]
    return df

# Generate and save both versions
df_v1 = generate_printing_data(num_samples=250, version=1)
df_v2 = generate_printing_data(num_samples=250, version=2)

os.makedirs('data', exist_ok=True)
df_v1.to_csv('data/printing_jobs_v1.csv', index=False)
df_v2.to_csv('data/printing_jobs_v2.csv', index=False)

print(f"V1 shape: {df_v1.shape}, V1 Warping class balance:\n{df_v1['warping_defect'].value_counts(normalize=True)}")
print(f"V2 shape: {df_v2.shape}, V2 Warping class balance:\n{df_v2['warping_defect'].value_counts(normalize=True)}")
```

```python
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

# Custom Feature Engineering Transformer
class FeatureEngineer(BaseEstimator, TransformerMixin):
    def __init__(self):
        pass
    def fit(self, X, y=None):
        return self
    def transform(self, X):
        X_copy = X.copy()
        # 1. Nozzle-Bed Temperature Difference
        X_copy['nozzle_bed_temperature_difference'] = X_copy['nozzle_temperature'] - X_copy['bed_temperature']
        # 2. Cooling Stress Index
        # More stress if fan speed is high and ambient temperature is low
        X_copy['cooling_stress_index'] = X_copy['cooling_fan_speed'] * (X_copy['printing_speed'] / (X_copy['ambient_temperature'] + 1e-5))
        return X_copy

# Test the pipeline assembly
numerical_cols = ['nozzle_temperature', 'bed_temperature', 'printing_speed', 'ambient_temperature',
                  'cooling_fan_speed', 'component_base_area', 'layer_height',
                  'nozzle_bed_temperature_difference', 'cooling_stress_index']

categorical_cols = ['material']

preprocessor = ColumnTransformer(
    transformers=[
        ('num', Pipeline([('imputer', SimpleImputer(strategy='median')), ('scaler', StandardScaler())]), numerical_cols),
        ('cat', Pipeline([('imputer', SimpleImputer(strategy='most_frequent')), ('onehot', OneHotEncoder(handle_unknown='ignore'))]), categorical_cols)
    ]
)

full_pipeline = Pipeline([
    ('engineer', FeatureEngineer()),
    ('preprocessor', preprocessor)
])

# Showcase transformation on V1
X_v1 = df_v1.drop(columns=['warping_defect'])
y_v1 = df_v1['warping_defect']

X_v1_trans = full_pipeline.fit_transform(X_v1)
print("Transformed features shape:", X_v1_trans.shape)
```

```python
!pip install -q mlflow

import mlflow
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, f1_score
from sklearn.pipeline import Pipeline

mlflow.set_tracking_uri("sqlite:///mlflow.db")
mlflow.set_experiment("3D_Printer_Warping_Defect")

X_train, X_test, y_train, y_test = train_test_split(X_v1, y_v1, test_size=30, random_state=42, stratify=y_v1)

trusted_types = [
    "__main__.FeatureEngineer",
    "numpy.dtype",
    "sklearn.compose._column_transformer._RemainderColsList",
    "sklearn.tree._tree.Tree"
]

with mlflow.start_run(run_name="Logistic_Regression") as run_lr:
    model_lr = Pipeline([
        ('pipeline', full_pipeline),
        ('classifier', LogisticRegression(max_iter=1000, class_weight='balanced'))
    ])
    model_lr.fit(X_train, y_train)
    preds = model_lr.predict(X_test)
    f1 = f1_score(y_test, preds)
    mlflow.log_metric("f1_score", f1)
    mlflow.sklearn.log_model(model_lr, "model", skops_trusted_types=trusted_types)
    print("Logistic Regression Report:\n", classification_report(y_test, preds))

best_rf_model = None
best_rf_f1 = -1
best_run_id = None

for i, (est, depth) in enumerate([(100, None), (150, 10), (200, 5)]):
    with mlflow.start_run(run_name=f"Random_Forest_Iter_{i+1}") as run_rf:
        model_rf = Pipeline([
            ('pipeline', full_pipeline),
            ('classifier', RandomForestClassifier(n_estimators=est, max_depth=depth, class_weight='balanced', random_state=42))
        ])
        model_rf.fit(X_train, y_train)
        preds_rf = model_rf.predict(X_test)
        f1_rf = f1_score(y_test, preds_rf)
        mlflow.log_metric("f1_score", f1_rf)
        mlflow.sklearn.log_model(model_rf, "model", skops_trusted_types=trusted_types)

        if f1_rf > best_rf_f1:
            best_rf_f1 = f1_rf
            best_rf_model = model_rf
            best_run_id = run_rf.info.run_id

print("Best Random Forest Iteration Report:\n", classification_report(y_test, best_rf_model.predict(X_test)))

if best_rf_f1 >= f1:
    mlflow.register_model(f"runs:/{best_run_id}/model", "WarpingPredictionModel")
else:
    mlflow.register_model(f"runs:/{run_lr.info.run_id}/model", "WarpingPredictionModel")
```
